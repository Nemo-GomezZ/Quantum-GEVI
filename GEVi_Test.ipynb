```python
import numpy as np
import matplotlib.pyplot as plt

# Parámetros
rho_crit = 1.0
Lambda_I = 1.0

def GEVI_F(I):
    return 1 - np.exp(-I)
```

```python
x = np.linspace(-10, 10, 500)

# densidad tipo función de onda
rho = np.exp(-x**2)

I = rho / rho_crit
F = GEVI_F(I)

# potenciales
phi_newton = -rho
phi_gevi = -rho * (1 + F)

plt.figure()
plt.plot(x, phi_newton, label="Newton")
plt.plot(x, phi_gevi, label="GEVI")
plt.title("Potencial: Newton vs GEVI (Gaussiana)")
plt.legend()
plt.show()
```

```python
x = np.linspace(-10, 10, 500)

# vacío + ruido
rho0 = 0.2
noise = 0.05 * np.random.normal(size=len(x))
rho = rho0 + noise

I = rho / rho_crit
F = GEVI_F(I)

phi = -rho * (1 + F)

plt.figure()
plt.plot(x, rho, label="Densidad (vacío + ruido)")
plt.plot(x, phi, label="Potencial GEVI")
plt.legend()
plt.title("Fluctuaciones cuánticas simuladas")
plt.show()
```

```python
x = np.linspace(-5, 5, 500)

# densidad muy alta en el centro
rho = 10 * np.exp(-x**2)

I = rho / rho_crit
F = GEVI_F(I)

plt.figure()
plt.plot(x, I, label="I(x)")
plt.plot(x, F, label="F(I) = saturación")
plt.legend()
plt.title("Saturación GEVI")
plt.show()
```

```python
r = np.linspace(0.1, 10, 500)

# perfil simple de masa
rho = 1 / (1 + r**2)

I = rho / rho_crit
F = GEVI_F(I)

# velocidad orbital ~ sqrt(r * dPhi/dr)
phi_newton = -rho
phi_gevi = -rho * (1 + F)

v_newton = np.sqrt(r * np.abs(np.gradient(phi_newton, r)))
v_gevi = np.sqrt(r * np.abs(np.gradient(phi_gevi, r)))

plt.figure()
plt.plot(r, v_newton, label="Newton")
plt.plot(r, v_gevi, label="GEVI")
plt.legend()
plt.title("Curva de rotación (toy model)")
plt.show()
```
